Question: Write a program code to print the comments to students of a Python coding course based on the marks as given in the table below.  Include appropriate comments in your program to help the user understand your program.

Code explanation: An if block is used to classify the user input into different ranges. Based on the range that the user input falls into, the respective comments will be returned at the end.

```python
# Requesting for user input on their marks
mark = input("Please enter your marks: ")

# Converting the variable mark from a string variable to an integer variable
mark = int(mark)

# Using an if block to determine the appropriate comments based on the user marks
if (mark >= 80):
  comments = "Excellent!"
elif (mark >= 70 and mark < 80):
  comments = "Very Good!"
elif (mark >= 60 and mark < 70):
  comments = "Good."
elif (mark >= 50 and mark < 60):
  comments = "Pass."
else:
  comments = "Fail. Please work harder."

# Printing out respective comments
print(comments)
```

Question: Write a program to write a for loop to print all the food dishes with a rating of at least 3 stars.

Code explanation: A for loop is used to check each dish and their number of stars. If they are 3 stars and more, the loop splits the item into the dish name before the '=' sign and the number of stars after the '=' sign. The dish name and star number will be appended in 2 lists, where they will be formatted and printed out.

```python
# Entering the required data
asian_food = ["nasi_lemak=**",
              "chicken_rice=***",
              "roti_chennai=***",
              "mohinga=****",
              "ru_rou_fan=***",
              "stinky_toufu=*",
              "peking_duck=***"]

# Creating an empty list to store dishes with at least 3 stars
threestar_dishes = []

# Creating an empty list to store the number of stars
stars = []

# Looping through to retrieve dishes with at least 3 stars
for i in range(len(asian_food)):
  if asian_food[i].count("*") >= 3:
    end_index = asian_food[i].index("=")
    threestar_dishes.append(asian_food[i][0:end_index])
    stars.append(asian_food[i][end_index+1:])

# Formatting the dish name to remove the underscore and capitalised first letter
for i in range(len(threestar_dishes)):
  threestar_dishes[i] = threestar_dishes[i].replace("_"," ")
  threestar_dishes[i] = threestar_dishes[i].title()

# Formatting and printing out dishes with at least 3 stars
print("{0:15} {1:15}".format("Dish", "Number of Stars"))
for i in range(len(threestar_dishes)):
  print("{0:15} {1:15}".format(threestar_dishes[i], stars[i]))
```

Question: Write a “Guess the number” program.

Code explanation: The code checks if the user input is a valid numeric input, and if it is within the valid range, and finally checks it against the generated magic number to reply to the user if they should guess higher or lower. When the magic number is guessed, the number of tries and the numbers guessed will be returned.

```python
# Importing the random package
import random

# Generating a random number between 1 and 500
n = random.randint(1,500)

# Creating an empty list to store user guesses
wrong_guesses = []

# Requesting for user number guesses
while True:
  guess = input("Please guess a postive number between 1 and 500: ")

# Checking if the user input is a valid number
  try:
    guess_int = int(guess)
  except ValueError:
    print("Invalid guess. Please guess a positive number between 1 and 500.")
    continue

# Checking if the user input is a valid guess between 1 and 500
  if (guess_int < 1 or guess_int > 500):
    print("Invalid guess. Please guess a positive number between 1 and 500.")
    print("."*80)

# Checking if the user input guessed the magic number and if not, should his guess be higher or lower
  elif (guess_int != n):
    wrong_guesses.append(guess_int)
    if (guess_int > n):
      print("{0} is incorrect. Please try a lower number.".format(guess_int))
      print("."*80)
    elif (guess_int < n):
      print("{0} is incorrect. Please try a higher number.".format(guess_int))
      print("."*80)
  else:
    wrong_guesses.append(guess_int)
    print("Congratulations! {0} is correct.".format(guess_int))
    break


# Formatting break
print("."*80)

# Printing out the list of guesses
print("You guessed the following numbers: {0}".format(wrong_guesses))

# Printing the number of tries before success
if len(wrong_guesses) == 1:
  print("You were successful in {0} try.".format(len(wrong_guesses)))
else:
  print("You were successful in {0} tries.".format(len(wrong_guesses)))
```

Question: Write a program to open the csv file called ‘nvidia_stock_prices.csv’ and generate 
a Report with the following details: 
- Number of days in the dataset
- Highest Price
- Lowest Price
- Median Price
- Average daily percentage change in price during the time period

Code explanation: A for loop is used to check through all the lines in the csv file and combine them into a new list after formatting. With the list of stock prices, the different statistics are calculated.

```python
# Importing libraries
import csv
import statistics as s

# Creating a variable to count total number of days and 2 empty lists to store the prices and percentage change between consecutive days
daycount = 0
pricelist = []
percentagechange = []

# Appending in the daily prices into pricelist list variable after converting it to a float variable and stripping the $ symbol
with open('nvidia_stock_prices.csv','r',newline='') as file:
  reader = csv.reader(file)
  for row in reader:
    if row[0] != "Date":
      daycount += 1
      pricelist.append(float(row[1].strip('$')))

# Calculating the highest, lowest and median price    
highestprice = max(pricelist)
lowestprice = min(pricelist)
medianprice = s.median(pricelist)

# Calculating the percentage change between days and appending it into the percentagechange list variable
for i in range(1,len(pricelist)):
    dailychange = ((pricelist[i] - pricelist[i-1]) / pricelist[i-1]) * 100
    percentagechange.append(dailychange)

# Calculating the average percentage change, rounded to 3 decimal places
averagepercentagechange = round(s.mean(percentagechange),3)

# Formatting and printing out the calculations
print("{0:^20} {1:^15}".format("Detail", "Calculation"))
print("{0:^20} {1:^15}".format("Total Days", daycount))
print("{0:^20} {1:^15}".format("Highest Price", highestprice))
print("{0:^20} {1:^15}".format("Lowest Price", lowestprice))
print("{0:^20} {1:^15}".format("Median Price", medianprice))
print("{0:^20} {1:^15}".format("Average % Change (%)", averagepercentagechange))
```
